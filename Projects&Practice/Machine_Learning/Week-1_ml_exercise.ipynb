```python
# collect 5 numbers and calculate average
print("Input 5 numbers to create the average")

number1 = float(input("input number 1: "))
number2 = float(input("input number 2: "))
number3 = float(input("input number 3: "))
number4 = float(input("input number 4: "))
number5 = float(input("input number 5: "))

print("Thank you. Calculating average...")

average = (number1 + number2 + number3 + number4 + number5) / 5
print(average)
```
