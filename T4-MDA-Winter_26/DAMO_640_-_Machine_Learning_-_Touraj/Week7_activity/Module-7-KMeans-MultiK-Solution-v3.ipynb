## Solution: Visualizing K-Means with Multiple k Values (v3)

This notebook shows how cluster boundaries evolve as k changes (2–6).

```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

X, y_true = make_blobs(n_samples=500, centers=4, n_features=5, random_state=42)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)

fig, axes = plt.subplots(2, 3, figsize=(12, 8))
axes = axes.ravel()
for i, k in enumerate([2, 3, 4, 5, 6]):
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    labels = km.fit_predict(X_scaled)
    axes[i].scatter(X_pca[:, 0], X_pca[:, 1], c=labels, cmap='tab10', s=20)
    axes[i].set_title(f'K-Means (k={k})')
    axes[i].set_xticks([])
    axes[i].set_yticks([])
plt.tight_layout()
plt.show()
```

**Observation:**
- At k=2, clusters are broad and overlapping.
- At k=3–4, clusters better match true structure.
- At k>4, clusters start splitting real groups unnecessarily.
