
## Opinion Structure in Sentiment Analysis

Opinions in sentiment analysis are often broken down into components such as **entity**, **aspect**, **sentiment**, **opinion holder**, and **time**. Using a product review example:

**Review Example:**
Posted by: *Sarah Lee*  
Date: *March 5, 2022*  
“(1) I bought the Galaxy ZX Pro smartphone three months ago.  
(2) I absolutely love using it.  
(3) Its display is crystal clear.  
(4) The battery easily lasts all day.  
(5) However, my brother says it feels too bulky.”

### Key Observations
- **Multiple Opinions:** Positive and negative sentiments about the same product.
- **Two People:** Sarah and her brother express opinions.
- **Time Stamp:** Review posted on March 5, 2022.

### Opinion Analysis
- Sentence (1): Objective statement (Neutral).
- Sentence (2): Positive opinion about the smartphone overall.
- Sentence (3): Positive sentiment about the display.
- Sentence (4): Positive sentiment about battery performance.
- Sentence (5): Negative opinion about size (bulkiness), attributed to another person.

### Formal Representation
An opinion can be represented as: Opinion = (g, s) <br>

Where:
- **g** = Target (entity or aspect, e.g., product, feature)
- **s** = Sentiment orientation (positive, negative, neutral) or numerical rating.

**Example:**
- Sentence (2): Target = *Galaxy ZX Pro*, Sentiment = *Positive*.


### Quintable notation:

Opinion = (eᵢ, aᵢⱼ, sᵢⱼₖₗ, hₖ, tₗ)

### Components:
- **eᵢ**: Entity (e.g., *Galaxy ZX Pro*)
- **aᵢⱼ**: Aspect of the entity (e.g., *display*, *battery*)
- **sᵢⱼₖₗ**: Sentiment expressed (e.g., *positive*, *negative*, *neutral*, or rating like *4 stars*)
- **hₖ**: Opinion holder (e.g., *Sarah*, *Sarah’s brother*)
- **tₗ**: Time when the opinion was expressed (e.g., *March 5, 2022*)

---

### Index Explanation Table

| **Index** | **Meaning**                                      | **Example**                                  |
|-----------|--------------------------------------------------|----------------------------------------------|
| i         | Identifies the entity                           | e₁ = Galaxy ZX Pro; e₂ = iPhone 15          |
| j         | Identifies the aspect of entity i              | a₁₁ = display; a₁₂ = battery; a₁₃ = camera  |
| k         | Identifies the opinion holder                  | h₁ = Sarah; h₂ = Sarah’s brother            |
| l         | Identifies the time of sentiment               | t₁ = March 5, 2022; t₂ = April 10, 2022     |
| sᵢⱼₖₗ    | Sentiment about aspect j of entity i by holder k at time l | positive, negative, neutral, or rating |

In this application activity, we will explore **sentiment data** and apply **classification techniques** using the `movie_reviews` dataset from **NLTK**—one of the most widely used natural language toolkits in Python.

4. **Model training and prediction**

```python
#imports and data loading
import pandas as pd
import os
import nltk
nltk.download('movie_reviews')
from nltk.corpus import movie_reviews
```

**If you are curious to see the whole dataset, you can download all reviews of this dataset as follows:**

```python
# Create main folder
base_dir = 'movie_review'
os.makedirs(base_dir, exist_ok=True)

# Save reviews to appropriate subfolders
for fileid in movie_reviews.fileids(): # iterating over all reviews in the NLTK movie_reviews.
    category = movie_reviews.categories(fileid)[0]  # extracting 'pos' or 'neg'
    review_text = movie_reviews.raw(fileid) # retrieving the full text of the review as a string

    # Create subfolder for category
    category_dir = os.path.join(base_dir, category) # building the folder path using base_dir
    os.makedirs(category_dir, exist_ok=True)

    # Save review using fileid
    filename = os.path.basename(fileid)
    filepath = os.path.join(category_dir, filename)

    # Write review text
    with open(filepath, 'w', encoding='utf-8') as f:
        f.write(review_text)

#You may want to save the reviews in a structured dataframe for further processing. 
docs = [(movie_reviews.raw(fileid), movie_reviews.categories(fileid)[0])
        for fileid in movie_reviews.fileids()]

moveis_review_df = pd.DataFrame(docs, columns=['review', 'label'])
print(moveis_review_df.head())
```

**Classification**

```python
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.metrics import classification_report
import random
```

**Preparing the documents.** <br>
In the following code snippet, the function movie_reviews.categories() returns the two sentiment labels available in the dataset: 'pos' for positive and 'neg' for negative. 
For each category, movie_reviews.fileids(category) retrieves the list of file IDs (unique identifiers for each review) that belong to that sentiment. Then, movie_reviews.raw(fileid) is used to load the full text content of a given review. 
By iterating through each file ID within both sentiment categories, you construct the variable _documents_ which is a list of tuples each containing the raw review text and its associated sentiment label. 

```python
documents = [(movie_reviews.raw(fileid), category)
             for category in movie_reviews.categories()
             for fileid in movie_reviews.fileids(category)]
```

**Shuffle the data to avoid any bias due to ordering.**

```python
random.shuffle(documents)
```

**Separate the data into features (text) and labels. Then Split into training and test sets.**

```python
texts = [text for text, label in documents] # list of review texts
labels = [label for text, label in documents] # list of corresponding labels ('pos' or 'neg')

X_train, X_test, y_train, y_test = train_test_split(
    texts, labels, test_size=0.2, random_state=42)
```

**Create a pipeline with TF-IDF and SVM.** <br>
Note that <br>
•	('tfidf', TfidfVectorizer(stop_words='english', ngram_range=(1, 2))): converts the reviews into numerical features using the TF-IDF method. <br>
•	 stop_words='english' removes common English words like “the”, “is”, “and” that carry little sentiment. <br>
•	ngram_range=(1, 2) tells it to include: Unigrams (e.g., "good", "excellent") and Bigrams (e.g., "very good", "not bad") <br>
•	'svm', LinearSVC()): applies Support Vector Machine (SVM) classifier with a linear kernel


```python
pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(stop_words='english', ngram_range=(1, 2))),
    ('svm', LinearSVC())
])
```

**Train the classifier.**

```python
pipeline.fit(X_train, y_train)
```

**Predict and evaluate.**

```python
y_pred = pipeline.predict(X_test)
print(classification_report(y_test, y_pred))
```

  - Weighted by the number of instances in each class.  
  - Example: Weighted F1 = (F1_neg × Support_neg + F1_pos × Support_pos) / Total_support  
  - Reflects overall classifier performance, accounting for class imbalance.

> Note that: The macro and weighted averages are equal here because the dataset is balanced (400 reviews total).
