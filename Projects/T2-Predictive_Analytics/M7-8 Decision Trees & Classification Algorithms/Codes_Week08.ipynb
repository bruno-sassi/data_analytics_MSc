```python
import numpy as np
import pandas as pd

# Create the training dataset
data = pd.DataFrame({
    'X1': [0, 2, 0, 0, -1, 1],
    'X2': [3, 0, 1, 1, 0, 1],
    'X3': [0, 0, 3, 2, 1, 1],
    'Y': ['Red', 'Red', 'Red', 'Green', 'Green', 'Red']
})

data
```

```python
# Test point
test_point = np.array([0, 0, 0])
```

```python
# Function to calculate Euclidean distance
def euclidean_distance(row, test_point):
    point = np.array([row['X1'], row['X2'], row['X3']])
    return np.sqrt(np.sum((point - test_point) ** 2))

# Calculate distances for each observation
data['Distance'] = data.apply(lambda row: euclidean_distance(row, test_point), axis=1)
```

```python
data
```

```python
# Sort by distance and add observation number
data['Observation'] = range(1, len(data) + 1)
```

```python
data
```

```python
result = data.sort_values('Distance')
result
```

```python
# Print results
print("\nEuclidean Distances from test point (0,0,0):")
print("=" * 60)
print("Obs  X1  X2  X3    Y     Distance")
print("-" * 60)
for idx, row in result.iterrows():
    print(f"{int(row['Observation']):2d}   {row['X1']:2.0f}   {row['X2']:2.0f}   {row['X3']:2.0f}   {row['Y']:<6} {row['Distance']:8.3f}")
print("=" * 60)

# Print ordered distances
print("\nOrdered from nearest to farthest:")
print("-" * 60)
for idx, row in result.iterrows():
    print(f"Observation {int(row['Observation'])}: distance = {row['Distance']:.3f} ({row['Y']})")
```
