# Dealing With Missing Values

```python
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf
```

```python
# Load and prepare the dataset (using a small dataset for demonstration)
data = {
 'Income':[56, 58, 60, 62, np.nan, 69, 73, 75],
 'Age':[22, 23, np.nan, 25, 27, 33, 36, 38],
 'Edu':[11, 13, np.nan, 14 , 15, 16, 18, 20]
}
```

```python
data
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
# Detect missing values
missing_values = df.isnull()
print("Missing Values:\n", missing_values)
```

```python
# Detect missing values
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
# Percentage of missing values in each column:
missing_percentage = (df.isnull().sum() / len(df)) * 100
print("Percentage of missing values in each column:")
print(missing_percentage[missing_percentage > 0])
```

```python
df_filled = df.fillna(0)
```

```python
print("\nDataFrame after replacing missing values with 0:")
print(df_filled)
```

```python
# Replace missing values with the mean of each column
df_mean_filled = df.fillna(df.mean())
print("\nDataFrame after replacing missing values with column means:")
print(df_mean_filled)
```

```python
# Replace missing values with the median of each column
df_median_filled = df.fillna(df.median())
print("\nDataFrame after replacing missing values with column medians:")
print(df_median_filled)
```

```python
df
```

```python
# Using forward fill to replace missing values
df_ffill = df.fillna(method='ffill')
print("\nDataFrame after forward filling missing values:")
print(df_ffill)
```

```python
# Using backward fill to replace missing values
df_bfill = df.fillna(method='bfill')
print("\nDataFrame after backward filling missing values:")
print(df_bfill)
```

```python
# Using interpolation to fill missing values
df_interpolated = df.interpolate()
print("\nDataFrame after interpolation of missing values:")
print(df_interpolated)
```

```python
# Prepare the formula for regression
# We'll predict column 'A' based on columns 'B' and 'C'
formula = 'Income ~ Age + Edu'
# Fit the regression model using only complete cases
model = smf.ols(formula, data=df.dropna()).fit()
```

```python
# Predict missing values
# Create a new DataFrame with missing values only
missing_data = df[df['Income'].isnull()]
```

```python
missing_data
```

```python
# Predict using the fitted model
predictions = model.predict(missing_data)
```

```python
predictions
```

```python
print("\nDataFrame after imputation:\n") # Imputation means estimating and filling the missing values in the dataset
df.loc[4, 'Income'] = predictions.values
print(df)
```

```python
# Create a sample DataFrame with missing values
data = {
'Income Level': ['High', 'Low', np.nan, 'Low', 'Low', np.nan, 'High'],
'Age': [20, 30, 40, 50, 60, 70, 80]
}
df = pd.DataFrame(data)
df
```

```python
# Detect missing values
missing_values = df['Income Level'].isnull()
print("Missing Values:\n", missing_values)
```

```python
# Count missing values
missing_count = df['Income Level'].isnull().sum()
print("\nMissing Value Count:", missing_count)
```

```python
# Calculate the mode of the 'Income Level' column
mode_category = df['Income Level'].mode()[0]
```

```python
mode_category
```

```python
# Replace missing values with the mode
df['Income Level'] = df['Income Level'].fillna(mode_category)
print("\nData after imputation:\n", df)
```

# Dealing With Outliers

```python
import numpy as np
import matplotlib.pyplot as plt
```

# Using Z-Score

```python
data = [10, 12, 12, 13, 12, 12, 100, 12, 14, 12, 12, 13]
mean = np.mean(data)
std_dev = np.std(data)
```

```python
mean
```

```python
std_dev
```

```python
# Detect outliers using Z-Score
threshold = 3
outliers_z_score = [x for x in data if abs((x - mean) / std_dev) > threshold]
```

```python
outliers_z_score
```

```python
# Visualization
plt.figure(figsize=(10, 5))
plt.plot(data, 'bo-', label='Data Points')
plt.axhline(y=mean + threshold * std_dev, color='r', linestyle='--',
label='Upper Threshold')
plt.axhline(y=mean - threshold * std_dev, color='r', linestyle='--',
label='Lower Threshold')
plt.title('Z-Score Method for Outlier Detection')
plt.xlabel('Index')
plt.ylabel('Values')
plt.legend()
plt.show()
```

# Using the IQR Method

```python
sample_data = np.array([1, 2, 2, 4, 5, 20])
```

```python
# Calculate IQR
q1 = np.percentile(sample_data, 25)
q3 = np.percentile(sample_data, 75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
```

```python
# Detect outliers
outliers_iqr = sample_data[(sample_data < lower_bound) | (sample_data > upper_bound)]
```

```python
outliers_iqr
```

```python
plt.figure(figsize=(10, 5))
plt.boxplot(sample_data)
plt.axhline(y=lower_bound, color='r', linestyle='--', label='Lower Bound')
plt.axhline(y=upper_bound, color='r', linestyle='--', label='Upper Bound')
plt.title('IQR Method for Outlier Detection')
plt.xlabel('Values')
plt.show()
```
