```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from statsmodels.tsa.stattools import adfuller

# Generate sample time series data
np.random.seed(0)
date_range = pd.date_range(start='1/1/2015', periods=108, freq='ME')
data = np.random.randn(108).cumsum()
time_series = pd.DataFrame(data, index=date_range)
time_series.head()
```

```python
# Plot the time series data
plt.figure(figsize=(10, 6))
plt.plot(time_series)
plt.title('Sample Time Series Data')
plt.xlabel('Date')
plt.ylabel('Value')
plt.grid(True)
plt.show()
```

```python
res = adfuller(time_series)
print("Test-Statistic:", res[0])
print("P-Value:", res[1])
if res[1] < 0.05:
  print("Stationary")
else:
  print("Non-Stationary")
```

# Augmented Dickey-Fuller (ADF) test.

```python
res = adfuller(x)
print("Test-Statistic:", res[0])
print("P-Value:", res[1])
if res[1] < 0.05:
  print("Stationary")
else:
  print("Non-Stationary")
```

```python
time_series['Diff'] = time_series.diff()
```

```python
time_series.head()
```

```python
time_series['Diff'].plot()
```

```python
res = adfuller(time_series['Diff'].dropna())
print("Test-Statistic:", res[0])
print("P-Value:", res[1])
if res[1] < 0.05:
  print("Stationary")
else:
  print("Non-Stationary")
```

```python
time_series['Diff_Diff'] = time_series['Diff'].diff()
```

```python
time_series.head()
```

```python
res = adfuller(time_series['Diff_Diff'].dropna())
print("Test-Statistic:", res[0])
print("P-Value:", res[1])
if res[1] < 0.05:
  print("Stationary")
else:
  print("Non-Stationary")
```
